# Standard Deviation and Variance

```python
%matplotlib inline
import numpy as np
import matplotlib.pyplot as plt

incomes = np.random.normal(100.0, 50.0, 10000)

plt.hist(incomes, 50)
plt.show()
```

```python
incomes.std()
```

```python
incomes.var()
```

## Activity

Experiment with different parameters on the normal function, and see what effect it has on the shape of the distribution. How does that new shape relate to the standard deviation and variance?

```python
# Let's experiment with different scale (standard deviation) values
import numpy as np
import matplotlib.pyplot as plt

# Scenario 1: Low Standard Deviation (Spread is narrow)
incomes_narrow = np.random.normal(100.0, 15.0, 10000)

# Scenario 2: High Standard Deviation (Spread is wide)
incomes_wide = np.random.normal(100.0, 75.0, 10000)

# Plotting both to compare
plt.figure(figsize=(12, 6))
plt.hist(incomes_narrow, 50, alpha=0.5, label='Narrow (std=15)', color='red')
plt.hist(incomes_wide, 50, alpha=0.5, label='Wide (std=75)', color='green')
plt.legend(loc='upper right')
plt.title('Effect of Standard Deviation on Distribution Shape')
plt.xlabel('Value')
plt.ylabel('Frequency')
plt.show()

print(f"Narrow Distribution - Realized Std: {incomes_narrow.std():.2f}, Variance: {incomes_narrow.var():.2f}")
print(f"Wide Distribution   - Realized Std: {incomes_wide.std():.2f}, Variance: {incomes_wide.var():.2f}")
```

##Observations and Relationship to Cybersecurity

1. Shape Change

A smaller standard deviation squeezes the data points closely around the mean, creating a tall, narrow peak.

A larger standard deviation spreads the data points far from the mean, producing a flatter, wider curve.

2. Mathematical Relationship

Standard deviation is the square root of the variance.

When standard deviation increases, variance increases quadratically.

3. Cybersecurity Connection

In cybersecurity anomaly detection, understanding this spread is critical. If your baseline system traffic has a narrow standard deviation, even a small spike in requests can easily stand out as a statistically significant anomaly. If the baseline has a wide standard deviation or high variance, detecting true threats from regular noisy traffic becomes much more difficult.

##Summary of parameters and distribution shape

When you decrease the standard deviation parameter scale, the curve becomes taller and narrower because the data points cluster tightly around the average. This represents a lower variance.

When you increase the standard deviation parameter, the curve flattens and widens because the data points spread out further from the average. This represents a higher variance.

In cybersecurity, a narrower distribution makes anomalous events much easier to spot, whereas a wider distribution can hide malicious activity in normal background noise.
